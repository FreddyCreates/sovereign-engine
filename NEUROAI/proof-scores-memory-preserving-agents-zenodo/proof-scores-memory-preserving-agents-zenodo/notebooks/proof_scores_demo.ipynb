# Proof Scores Demo

This notebook demonstrates the expanded seed-suite scoring shape for the Proof Scores benchmark proposal. It is not a validated benchmark run.

```python
import json
from pathlib import Path
import sys

ROOT = Path('..').resolve()
sys.path.append(str(ROOT / 'code'))
from proofscore import evaluate_against_minimums

tasks = [json.loads(line) for line in (ROOT / 'data' / 'sample_tasks.jsonl').read_text().splitlines()]
families = sorted({task['corpus_family'] for task in tasks})
len(tasks), families
```

```python
task = tasks[0]
toy_observation = {dimension: max(2, task.get('expected_min_scores', {}).get(dimension, 2)) for dimension in [
    'claim_score', 'evidence_score', 'boundary_score', 'lineage_score', 'promotion_score', 'notary_readiness'
]}

task['task_id'], task['corpus_family'], evaluate_against_minimums(toy_observation, task['expected_min_scores'])
```
